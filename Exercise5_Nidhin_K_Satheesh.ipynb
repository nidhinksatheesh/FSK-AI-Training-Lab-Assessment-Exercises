### Install & Import Required Libraries

```python

import time
import gradio as gr
```

### Task 1: Define Conversational Response Logic

```python
def ai_assistant_response(message: str, history: list) -> str:
    """
    Generates contextual conversational responses for the Gradio ChatInterface.
    """
    user_text = message.strip().lower()
    turn_number = len(history) // 2 + 1 if history else 1

    if "pandas" in user_text or "dataframe" in user_text:
        reply = (
            "**Pandas Tip:** Use `df.groupby('col').agg(...)` for clean group-level summaries "
            "and `df.to_excel('output.xlsx', index=False)` to export clean spreadsheets."
        )
    elif "opencv" in user_text or "blend" in user_text or "image" in user_text:
        reply = (
            "**OpenCV Image Blending:** Ensure both images share identical `(height, width)` via "
            "`cv2.resize()`, then blend them linearly with `cv2.addWeighted(img1, 0.6, img2, 0.4, 0)`."
        )
    elif "gradio" in user_text or "interface" in user_text:
        reply = (
            "**Gradio Prototyping:** `gr.ChatInterface` automatically manages chat history, "
            "sample prompt buttons, and streaming responses in just a few lines of Python."
        )
    elif "summarize" in user_text or "history" in user_text:
        reply = f"We are currently on **Turn {turn_number}** of our conversation, and I am actively tracking your session context!"
    else:
        reply = (
            f"You asked: *\"{message.strip()}\"*\n\n"
            "I am a lightweight Gradio AI prototype assistant. Try asking me about **Pandas**, "
            "**OpenCV image blending**, **Gradio UI prototyping**, or click one of the sample prompts below!"
        )

    # Simulate natural token streaming
    partial = ""
    for word in reply.split(" "):
        partial += word + " "
        time.sleep(0.03)
        yield partial.strip()
```

### Task 2: Configure & Launch the Gradio `ChatInterface`
Configure `gr.ChatInterface` with:
- A clear **Title** (`AI Lab Prototyping Assistant`)
- A **Welcome / Usage Description** explaining how to interact with the chatbot
- **Four Sample Prompts (`examples`)** so users can click and test responses immediately in the browser
- `demo.launch()` (renders inline inside Jupyter and opens a local web URL at `http://127.0.0.1:7860`)

```python
sample_prompts = [
    "How do I blend two images using OpenCV in Python?",
    "Give me a quick tip for aggregating data in Pandas.",
    "Why is Gradio useful for rapid AI prototyping?",
    "Summarize our conversation history so far."
]

demo = gr.ChatInterface(
    fn=ai_assistant_response,
    title="AI Lab Prototyping Assistant",
    description=(
        "Welcome! Type a message below or click any of the sample prompts to test real-time "
        "conversational AI responses directly in your browser."
    ),
    examples=sample_prompts,
    cache_examples=False
)

# Launch inline in Jupyter and expose local server URL
demo.launch(inline=True, share=False)
```
