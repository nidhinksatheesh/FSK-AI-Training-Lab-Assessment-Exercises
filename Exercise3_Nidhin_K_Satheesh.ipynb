### 2. Exact Structured Video Synthesis Prompt

> **Prompt Text:**
> *"Generate a strict 3-second video clip (exact total duration: 3 seconds, no longer) that smoothly transitions between the three uploaded images in sequence:

- **0s–1s:** Start on Image 1 (the grey-and-white husky statue with a red scarf in the green park) with subtle natural light.
- **1s–2s:** Smoothly cross-dissolve through Image 2 (the blended double-exposure composite of the husky statue and the bear hot air balloon).
- **2s–3s:** Resolve cleanly onto Image 3 (the cartoon bear hot air balloon with a red bow floating in the clear blue sky) and end at 3 seconds. Camera: Gentle upward tilt with steady framing. Keep subjects sharp, natural lighting, smooth dissolve, no distortion, and stop the clip strictly at 3 seconds."*

---

### 3. Camera, Motion & Transition Instructions
* **Camera Movement:** Gentle upward tilt (`pan-up`) combined with a subtle `3%` slow dolly-in, guiding the viewer's eye naturally from the grounded husky statue in the park (`0.0s`) toward the open sky and ascending bear hot air balloon (`3.0s`).
* **Transition Curve & Pacing:**
  * `0.0s – 1.0s`: Hold on Source Image 1 (`output_original_1.png`) with subtle leaf sway and natural specular highlights on the glossy husky sculpture.
  * `1.0s – 2.0s`: Smooth luminance-matched cross-dissolve passing directly through the weighted composite (`output_blended_composite.png`) at the `1.5s` midpoint, harmonizing the red bandana and red balloon bow accents.
  * `2.0s – 3.0s`: Clean visual resolve into Source Image 2 (`output_original_2.png`), capturing gentle buoyant drift of the bear hot air balloon against the clear blue sky.
* **Negative Constraints (Quality Guardrails):** Avoid facial distortion on the statue or balloon, flickering sky gradients, unnatural double limbs/ears, abrupt jump cuts, or camera shake.
