## Exercise 1: Exploratory Data Analysis & Automated Reporting with Pandas

Objective: Demonstrate programmatic data generation, statistical aggregation,
visualization and spreadsheet export.

### Task Description:
1. Create a Pandas DataFrame containing four columns—f1, f2, f3, and f4—
populated with random numerical values.
2. Generate a scatter plot illustrating the relationship between column f2 and
column f3 (include appropriate axis labels and a chart title).
3. Compute the row-wise arithmetic mean across f1, f2, f3, and f4 and store the
result in a new column named f5.
4. Export the complete DataFrame (containing all five columns: f1 through f5) to
a Microsoft Excel file (.xlsx).

### Import required libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

%matplotlib inline
```

### Task 1: Create the DataFrame and populate with random values

```python

np.random.seed(42)

# No of rows=100
#No of columns=4
num_rows = 100
data = np.random.uniform(low=0.0, high=100.0, size=(num_rows, 4))

# Create the DataFrame
df = pd.DataFrame(data, columns=['f1', 'f2', 'f3', 'f4'])
```

### Task 2: Generate a Scatter Plot (f2 vs. f3)

```python
plt.figure(figsize=(8, 5))
plt.scatter(df['f2'], df['f3'], color='steelblue', edgecolor='black', alpha=0.75)

# Add chart title and axis labels
plt.title('Scatter Plot: Relationship Between f2 and f3', fontsize=14, fontweight='bold')
plt.xlabel('Feature f2', fontsize=12)
plt.ylabel('Feature f3', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.6)

plt.tight_layout()
plt.show()
```

### Task 3: Compute Row-Wise Arithmetic Mean and add it in column f5

```python
# Compute row-wise arithmetic mean across f1 through f4
df['f5'] = df[['f1', 'f2', 'f3', 'f4']].mean(axis=1)
```

### Task 4: Export Complete DataFrame to Excel (.xlsx)

```python
output_filename = 'report_output.xlsx'

# Export DataFrame to Excel using openpyx1
df.to_excel(output_filename, index=False, sheet_name='Report')

print(f"DataFrame successfully exported to '{output_filename}' with shape {df.shape}.")
```
